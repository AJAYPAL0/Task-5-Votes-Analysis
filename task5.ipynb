```python
df = pd.read_csv("1976-2020-president.csv")
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
highest_votes = df.loc[df["totalvotes"].idxmax()]

print(highest_votes)
```

```python
lowest_votes = df.loc[df["totalvotes"].idxmin()]

print(lowest_votes)
```

```python
top_votes = df.sort_values(by="totalvotes", ascending=False).head(10)

print(top_votes[["candidate", "totalvotes"]])
```

```python
plt.figure(figsize=(10,5))

sns.barplot(x="totalvotes", y="candidate", data=top_votes)

plt.title("Top 10 Candidates by Votes")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="totalvotes", y="candidatevotes", data=df)

plt.title("Total Votes vs Candidate Votes")

plt.show()
```

```python
correlation = df["totalvotes"].corr(df["candidatevotes"])

print("Correlation:", correlation)
```
